### Barplots from aggregated values

```python
from sklearn import datasets
```

```python
iris=datasets.load_iris() #from sklearn 
iris=pd.DataFrame(data=np.c_[iris['data'],iris['target']],columns=iris['feature_names']+['species'])
iris.species = np.where(iris.species == 0.0, 'setosa', np.where(iris.species == 1.0,'versicolor','virginica'))
iris.columns = iris.columns.str.replace(' ','')
```

```python
iris.head(5)
```

```python
isa=iris.groupby(['species']).mean() #average of different measurements across species
```

```python
isa
```

```python
sd=iris.groupby(['species']).std() 
sd #standard deviation represents the variance in the data
```

```python
my_plot = isa.plot(kind='bar') #barplot of the average meaurement values for the 3 species
my_plot = isa.plot()
my_plot.set_ylabel("Average values")
```

```python
isa.plot(kind='bar', yerr=sd) #use sd as error bar
#Error bars are graphical representations of the variability of data 
#used on graphs to indicate the error or uncertainty in a reported measurement (such as mean)
```

## Pie Chart
### a type of graph in which a circle is divided into sectors that each represent a proportion of the whole.

```python
%matplotlib inline

import matplotlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
d=pd.read_csv(r"D:\UNEC\python_data analytics\influence_final.csv") #aggregate influence of all the universities in a country
```

```python
d.head(4)
```

```python
d1=d[:8]
```

```python
d1
```

```python
# Create a pie chart
plt.pie(
    # influence of universities
    d1['InfluenceAggregate'],
    # by country
    labels=d1['country'])
```

```python
colors = ["red", "yellow", "lightskyblue", "blue", "green", "pink", "purple"]


plt.pie(d1['InfluenceAggregate'],labels=d1['country'], colors=colors)

plt.axis('equal')
plt.show()   
   
```

```python
colors = ["#E13F29", "#D69A80", "#D63B59", "#AE5552", "#CB5C3B", "#EB8076", "#96624E"]


plt.pie(d1['InfluenceAggregate'],labels=d1['country'], colors=colors,autopct='%1.0f%%')

plt.axis('equal')
plt.show()  
```

#### Time Series of rainfall data from the different subdivisions of India (1901-2015)

```python
rain1=pd.read_csv(r"D:\UNEC\python_data analytics\rainIndia.csv")
```

```python
rain1.head(4)
```

```python
rain1.SUBDIVISION.unique() #list the sub-divisions in India
```

```python
andnico=rain1[rain1.SUBDIVISION == "ANDAMAN & NICOBAR ISLANDS"]
```

```python
andnico.head(6)
```

```python
plt.plot(andnico['YEAR'],andnico['ANNUAL'])
plt.ylabel('Annual Rainfall')
plt.xlabel('Year')
plt.title("Andaman & Nicobar Rainfall Time Series")
plt.show()
```

```python
x=rain1.loc[rain1['SUBDIVISION'].isin(['ANDAMAN & NICOBAR ISLANDS','CHHATTISGARH','TAMIL NADU','TELANGANA','EAST RAJASTHAN'])]
```

```python
x.tail(6)
```

```python
x.SUBDIVISION.unique()
```

```python
plt.plot(x['YEAR'],x['ANNUAL'])
plt.ylabel('Annual Rainfall')
plt.xlabel('Year')
plt.title("India Rainfall Time Series")
plt.show()
```

```python
annico=x[x.SUBDIVISION == "ANDAMAN & NICOBAR ISLANDS"]
ch=x[x.SUBDIVISION == "CHHATTISGARH"]
er=x[x.SUBDIVISION == "EAST RAJASTHAN"]
tn=x[x.SUBDIVISION == "TAMIL NADU"]
t=x[x.SUBDIVISION == "TELANGANA"]
```

```python
fig = plt.figure(figsize=(15, 15)) #size of plot
plt.plot(annico['YEAR'],annico['ANNUAL'], label='Andaman')
plt.plot(ch['YEAR'],ch['ANNUAL'], marker='o', linestyle='--', color='r', label='Chattisgarh')
plt.plot(er['YEAR'],er['ANNUAL'], marker='o', linestyle='--', color='g', label='E Rajasthan')
plt.plot(tn['YEAR'],tn['ANNUAL'], marker='o', linestyle='--', color='y', label='TN')

plt.ylabel('Annual Rainfall')
plt.xlabel('Year')
plt.title("India Rainfall Time Series")
plt.legend()
plt.show()
```
