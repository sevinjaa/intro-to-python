The multiplication operator * for arrays performs element-wise multiplication, not matrix multiplication.
Arrays are the preferred data structure in NumPy for most applications due to their flexibility and support for a wide range of mathematical and scientific operations.

Matrix:
The numpy.matrix class is a specialized 2D array primarily used for linear algebra operations.
Matrix objects are always 2-dimensional, whereas arrays can have any number of dimensions.
The multiplication operator * is defined for matrix objects to perform matrix multiplication (dot product).
Matrix objects have some additional methods for linear algebra operations, such as I for matrix inversion and T for matrix transposition.
Matrix objects are less commonly used in NumPy and are considered somewhat deprecated in favor of arrays for most applications.

Set Operations with np.array data type

```python
s1=np.array(['desk','chair','bulb'])
s2=np.array(['lamp','bulb','chair'])

print(np.intersect1d(s1,s2))
print(np.union1d(s1,s2))
print(np.setdiff1d(s1,s2))
print(np.setdiff1d(s2,s1))
print(np.in1d(s1,s2))
```

NORMAL DISTRIBUTION

Use the random.normal() method to get a Normal Data Distribution.
It has three parameters:
loc - (Mean) where the peak of the bell exists.
scale - (Standard Deviation) how flat the graph distribution should be.
size - The shape of the returned array.

```python
import matplotlib.pyplot as plt
import seaborn as sns
dist_n=random.normal(loc=2, scale=1, size=(100))

print(dist_n)

sns.displot(dist_n, kind='kde')
plt.show()
```

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns 
dist_n = random.normal(loc=2, scale=1, size=(100))

print(dist_n)
# plt.hist(dist_n)

sns.distplot(dist_n, hist=False)
plt.show()
```

Binomial Distribution

Binomial Distribution is a Discrete Distribution.
It has three parameters:
n - number of trials.
p - probability of occurence of each trial (e.g. for toss of a coin 0.5 each).
size - The shape of the returned array.

```python
from numpy import random
dist_b=random.binomial(n=20, p=0.4, size=10)

print(dist_b)
```

```python
from numpy import random
dist_b = random.binomial(n=20, p=0.4, size=10)
print(dist_b)

sns.distplot(dist_b, hist=True, kde=True) #kernel density estimation (kde)
plt.show()
```
